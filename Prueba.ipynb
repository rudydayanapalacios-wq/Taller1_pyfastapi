##**Sistema de registro de Aprendices SENA**
Proeycto en Python para la gestión  de aprendicers mediante intrerfaz por consola

Registar la siguiente informacion mediante una funcion:

°Tipo de doc

°documento

°nombre

°apellidos

°ficha

°Programa

```python


#COMO LO HIXO EL PROFE

aprendices = [] # Initialize the list of apprentices

def registrar_aprendiz():
  print("Registre el aprendiz")
  documento=input("numero de documento: ").strip()

  #validar si el aprendiz ya existe
  for a in aprendices:
     if a["documento"] == documento:
       print("El aprendiz ya existe")
       return

  tipo_doc=input("Tipo de documento(CC/TI/CE) ").strip().upper()
  nombre=input("Nombre: ").strip().upper()
  ficha=input("fICHAS: ").strip()
  programa=input("programa de formacion:").strip().title()

  #guardar aprendiz
  nuevo_aprendiz={
      "tipo_doc": tipo_doc,
      "documento": documento,
      "nombre": nombre,
      "ficha": ficha,
      "programa": programa
  }

  aprendices.append(nuevo_aprendiz)
  print("Aprendiz registrado exitosamente!")

# Example usage (uncomment to test):
# registrar_aprendiz()
# print(aprendices)
```
